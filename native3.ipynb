###  Class Native Transformer(fixing neutral problem)

```python
import pandas as pd
from transformers import pipeline
from sklearn.metrics import classification_report, confusion_matrix

pd.set_option("display.max_colwidth",150)

print("\n Loading 3-class native sentimental model(cardiffnlp/twitter-roberta-base-sentiment)")
sentiment_model = pipeline("sentiment-analysis",model="cardiffnlp/twitter-roberta-base-sentiment")

label_map ={"LABEL_0":"negative","LABEL_1":"neutral","LABEL_2":"positive"}
def native_3class_sentiment(text):
    result = sentiment_model(text[:512])[0]
    return label_map[result["label"]]

# check on 1000-review samples and compare with previous transformer performance
sample = pd.read_csv("C:/Users/Priyanka/Datascience/NLP_projects/transformer_full_sample.csv")

#Running 3-class native model on the same 1000 reviews..")
sample["native3_pred"]=sample["clean_text"].apply(native_3class_sentiment)

print("\n====================================================================")
print("\n 3-class Native Model - Classification Report")
print("\n====================================================================")
print(classification_report(sample["sentiment"],sample["native3_pred"]))

cm = confusion_matrix(
    sample["sentiment"], sample["native3_pred"],
    labels=["negative", "neutral", "positive"]
)
print("Confusion Matrix (rows=actual, cols=predicted):")
print(pd.DataFrame(cm, index=["neg", "neu", "pos"], columns=["neg", "neu", "pos"]))
```

```python
sample.to_csv("native3_transformer.csv", index=False)
print("Saved successfully")
```
